```python
## Exploratory Data Analysis (EDA)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
sns.set(style="whitegrid")
```

```python
df = pd.read_csv("default of credit card clients.csv", skiprows=1)
df.head()
```

```python
### Verifying Column Names
print(df.columns.tolist())
```

```python
### Cleaning Column Names
print("Shape of dataset:", df.shape)
print("\nColumn names:\n", df.columns.tolist())
df.info()
```

```python
### Cleaning Column Names
df.rename(columns={"default payment next month": "default"}, inplace=True)
df.drop(columns=["ID"], inplace=True)
df.head()
```

```python
### Checking for Missing Values
print("Missing values per column:\n")
print(df.isnull().sum())
print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
### Checking for Duplicate Rows
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df.describe()
```

```python
### Target Variable Distribution
print(df['default'].value_counts())
print("\nPercentage:\n", df['default'].value_counts(normalize=True) * 100)

plt.figure(figsize=(5,4))
sns.countplot(x='default', data=df)
plt.title("Distribution of Default Payment (Target Variable)")
plt.xlabel("Default (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.show()
```

```python
### Distribution of Numeric Features
numeric_cols = ['LIMIT_BAL', 'AGE', 'BILL_AMT1', 'PAY_AMT1']
df[numeric_cols].hist(figsize=(10,8), bins=30)
plt.suptitle("Distribution of Key Numeric Features")
plt.tight_layout()
plt.show()
```

```python
### Distribution of Categorical Features
categorical_cols = ['SEX', 'EDUCATION', 'MARRIAGE']
fig, axes = plt.subplots(1, 3, figsize=(15,4))
for i, col in enumerate(categorical_cols):
    sns.countplot(x=col, data=df, ax=axes[i])
    axes[i].set_title(f"Distribution of {col}")
plt.tight_layout()
plt.show()
```

```python
### Outlier Detection
plt.figure(figsize=(12,6))
sns.boxplot(data=df[['LIMIT_BAL', 'BILL_AMT1', 'PAY_AMT1', 'AGE']])
plt.title("Boxplot for Outlier Detection")
plt.xticks(rotation=45)
plt.show()
```

```python
### Correlation Heatmap
plt.figure(figsize=(16,12))
corr = df.corr()
sns.heatmap(corr, annot=False, cmap='coolwarm')
plt.title("Correlation Heatmap of Features")
plt.show()
```

```python
### Feature Correlation with Target
target_corr = df.corr()['default'].sort_values(ascending=False)
print(target_corr)
```

```python
### Bivariate Analysis: Key Features vs Target
plt.figure(figsize=(7,5))
sns.boxplot(x='default', y='LIMIT_BAL', data=df)
plt.title("Credit Limit vs Default Status")
plt.show()

plt.figure(figsize=(7,5))
sns.boxplot(x='default', y='AGE', data=df)
plt.title("Age vs Default Status")
plt.show()
```
